### Reshaping, stacking, squeezing and unsqueezing tensors

```python
# Create tensor
import torch
x = torch.arange(1., 10.)
x, x.shape
```

```python
# Change the dimensions
x_reshaped = x.reshape(3, 3)
x_reshaped, x_reshaped.shape
```

```python
# Change the view
z = x.view(1, 9)
z, z.shape
```

```python
# Changing z changes x (because a view of a tensor shares same memory as the orginal tensor)
z[:, 0] = 5
z, x
```

```python
# Stacking tensors on top of each other
x_stacked = torch.stack([x, x, x, x], dim=0)
x_stacked, x_stacked.shape
```

```python
# Squeezing -> torch.squeeze() - removes all single dimensions from a target tensor
x = torch.zeros(2, 1, 2, 1, 2)
x, x.shape
```

```python
x_squeezed = x.squeeze()
x_squeezed, x_squeezed.shape
```

```python
y = x.squeeze(0)
y, y.shape
```

```python
y = x.squeeze(1)
y, y.shape
```

```python
y = x.squeeze(2)
y, y.shape
```

```python
y = x.squeeze(4)
y, y.shape
```

```python
# torch.unsqueeze() - adds a single dimension to a target tensor at a specific dim (dimension)
print(x_squeezed, x_squeezed.shape)
x_unsqueezed = x_squeezed.unsqueeze(dim=1)
print(x_unsqueezed, x_unsqueezed.shape)
```

```python
# torch.permute() - rearanges the dimensions of the tensor
x = torch.randn(2, 3, 5)
x.shape
```

```python
torch.permute(x, (2,0,1)).size()
```

```python
original_image = torch.rand(size=(244, 244, 3)) # [height, width, color-channel]

rearanged_image = original_image.permute(2, 0, 1).shape
rearanged_image# [color-channel, height, width]
```

## Pytorch tensors and NumPy

NumPy is a popular scientific Python numerical computing library.

And because of this, PyTorch has functionality to interact with it.

* Data in NumPy, want in PyTorch tensor -> `torch. from_numpy(ndarray)`
* PyTorch tensor -> NumPy -> `torch.Tensor.numpy`

```python
# NumPy array to PyTorch tensor
import torch
import numpy as np

array = np.arange(1.0, 8.0)
tensor = torch.from_numpy(array)
array, tensor
```

```python
# PyTorch tensor to NumPy array
tensor = torch.ones(7)
array = tensor.numpy()
tensor, array
```

### Reproducibility (trying to take random out of random)

```python
import torch

# Create two random tensors
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

```python
# Let's create some random but reproducible tensors

# Set a random seed
RANDOM_SEED = 42

torch.manual_seed(RANDOM_SEED)
random_tensor_C = torch.rand(3, 4)

torch.manual_seed(RANDOM_SEED)
random_tensor_D = torch.rand(3, 4)

print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D)
```
